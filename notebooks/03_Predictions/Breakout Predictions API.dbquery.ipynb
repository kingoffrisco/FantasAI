```sql
-- Fantasy Breakout Predictions API Endpoint
-- Returns top waiver wire targets with usage signals and news sentiment
-- Updated every Tuesday after MNF via scheduled job

SELECT 
  player_name,
  position,
  team,
  season,
  week,
  -- Usage metrics
  snap_share as snap_percentage,
  snap_share_delta as snap_pct_change,
  touches as total_touches,
  fantasy_points as last_week_points,
  opportunity_score,
  avg_snap_share_prev_2wk as avg_snap_pct_2wk,
  avg_fantasy_points_prev_2wk as avg_points_2wk,
  -- Composite breakout score (0-1 scale)
  breakout_score,
  alert_level,
  -- News sentiment context
  news_sentiment,
  news_impact_score,
  news_volume as news_mentions,
  has_opportunity_news as opportunity_change_mentioned,
  has_injury_news as injury_concern,
  news_buzz_score,
  -- Metadata
  generated_at as prediction_timestamp,
  -- Ranking
  ROW_NUMBER() OVER (ORDER BY breakout_score DESC) as rank
FROM main.fantasai.breakout_predictions_current
WHERE breakout_score > 0.05  -- Filter to meaningful signals
ORDER BY breakout_score DESC
LIMIT 50
```

# FantasAI Cloudflare Worker API Test

This notebook tests all endpoints from your Cloudflare Worker to verify they're returning data correctly.

```python
import requests
import json
from pyspark.sql import Row
from pyspark.sql import functions as F

BASE_URL = "https://api.fantasai.net"
```

```python
# Test injuries endpoint
print("Testing /api/v1/injuries")
try:
    response = requests.get(f"{BASE_URL}/api/v1/injuries", timeout=30)
    print(f"Status: {response.status_code}")
    
    if response.status_code == 200:
        data = response.json()
        print(f"✓ Success! Records: {len(data) if isinstance(data, list) else 'N/A'}")
        if isinstance(data, list) and len(data) > 0:
            print(f"Sample keys: {list(data[0].keys())}")
            print(f"First record: {json.dumps(data[0], indent=2)}")
    else:
        print(f"Error: {response.text}")
except Exception as e:
    print(f"Error: {e}")
```
